```python
import os
import glob
def load_fewsnet_shapefiles(root_dir, countries):
    records = []
    for path in glob.glob(os.path.join(root_dir, "**", "*_CS.shp"), recursive=True):
        fname = os.path.basename(path).replace("_CS.shp", "")
        region, year, month = fname.split("_")[0], int(fname.split("_")[1][:4]), int(fname.split("_")[1][4:6])

        gdf = gpd.read_file(path)
        gdf["region"], gdf["year"], gdf["month"] = region, year, month
        if gdf.crs is None:
            gdf = gdf.set_crs("EPSG:4326")
        gdf = gdf.to_crs(countries.crs)

        joined = gpd.sjoin(gdf, countries, how="left", predicate="intersects")
        records.append(joined)
    return pd.concat(records, ignore_index=True)
```

```python
def fetch_country_boundaries():
    url = "https://raw.githubusercontent.com/nvkelso/natural-earth-vector/master/geojson/ne_110m_admin_0_countries.geojson"
    return gpd.read_file(url)[["NAME", "ADM0_A3", "geometry"]]

def join_to_countries(gdf, countries):
    return gpd.sjoin(gdf.to_crs(countries.crs), countries, how="left", predicate="intersects")
```
